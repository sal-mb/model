# Knapsack Problem Exercises
This notebook uses the Julia programming language to solve the exercises, the cells bellow will explain how to install the Julia kernel and packages to use with this notebook.

```julia
using JuMP
import HiGHS
using Combinatorics

include("check_exercise.jl");
```

```julia
n = 7;
capacity = 50;

profits = [70, 20, 39, 37, 7, 5, 10];
weights = [31, 10, 20, 19, 4, 3, 6];
```

```julia
function knapsack_ex1(n::Int, capacity::Int, profits::Vector{Int}, weights::Vector{Int}, test::Vector{Int})
    
    tests = generate_partitions(n);
    
    j = 1
    k = 2

    model = Model(HiGHS.Optimizer);
    
    @variable(model, x[1:n], Bin);
    set_silent(model);
    @constraint(model, sum(weights[i] * x[i] for i in 1:n) <= capacity);
    #Add your constraints here
    @constraint(model, x[j] <= x[k])
    
    
    @objective(model, Max, sum(profits[i] * x[i] for i in 1:n));

    set_test(model, test, n, x);
    
    optimize!(model);
    return termination_status(model);
end;

check_exercise(knapsack_ex1, n, capacity, profits, weights, 1)
```

```julia
function knapsack_ex2(n::Int, capacity::Int, profits::Vector{Int}, weights::Vector{Int}, test::Vector{Int})
    
    tests = generate_partitions(n);
    
    j = 1
    k = 2

    model = Model(HiGHS.Optimizer);
    
    @variable(model, x[1:n], Bin);
    set_silent(model);
    @constraint(model, sum(weights[i] * x[i] for i in 1:n) <= capacity);
    #Add your constraints here
    @constraint(model, x[j] + x[k] <= 0.5)
    
    
    @objective(model, Max, sum(profits[i] * x[i] for i in 1:n));

    set_test(model, test, n, x);
    
    optimize!(model);
    return termination_status(model);
end;

check_exercise(knapsack_ex2, n, capacity, profits, weights, 2)
```

```julia
function knapsack_ex3(n::Int, capacity::Int, profits::Vector{Int}, weights::Vector{Int}, test::Vector{Int})
    
    tests = generate_partitions(n);
    
    j = 1
    k = 2
    l = 3
    model = Model(HiGHS.Optimizer);
    
    @variable(model, x[1:n], Bin);
    set_silent(model);
    @constraint(model, sum(weights[i] * x[i] for i in 1:n) <= capacity);
    #Add your constraints here
    @constraint(model, x[l] <= x[j])
    @constraint(model, x[l] <= x[k])
    @constraint(model, x[l] >= x[j] + x[k] - 1)
    
    @objective(model, Max, sum(profits[i] * x[i] for i in 1:n));

    set_test(model, test, n, x);
    
    optimize!(model);
    return termination_status(model);
end;

check_exercise(knapsack_ex3, n, capacity, profits, weights, 3)
```
