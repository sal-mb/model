### Exercise 1
Modify the KP formulation to address a case where two arbitrary items $j,k \in \{1,\dots,n\}$ are linked so that if $j$ is included in the knapsack, so must be $k$. In other words, solutions where $x_j = 1$ and $x_k = 0$ must not be allowed.
